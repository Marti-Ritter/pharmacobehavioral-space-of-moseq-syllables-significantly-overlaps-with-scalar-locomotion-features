```python
import numpy as np
import pandas as pd
import pathlib
import pickle
from behavioral_analysis.utility.builtin_classes.objects import save_object
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
root_dir = pathlib.Path(r"H:\2026_Letter")

data_root_dir = root_dir / "data"
code_root_dir = root_dir / "code"

data_output_dir = code_root_dir / "data_output"
data_output_dir.mkdir(exist_ok=True, parents=True)
```

```python
with open(data_root_dir / "fingerprints.pkl", "rb") as f:
    fingerprints, fingerprint_labels = pickle.load(f, encoding='latin1')

fingerprint_labels
```

```python
drug_order = [(x if x!="control" else "VEH") for x in fingerprint_labels["unique_drug"]]  # update to new schema
with open(data_output_dir / "drug_order.txt", mode="w") as f:
    f.writelines([s + "\n" for s in drug_order])
```

```python
index_levels = ["drug_class", "drug", "highorlow", "mouse_names"]
index = list(zip(*[fingerprint_labels[x] for x in index_levels]))

scalars_df = pd.DataFrame(data=fingerprints["scalars"], index=index)
scalars_df.index = pd.MultiIndex.from_tuples(scalars_df.index, names=index_levels)

moseq_df = pd.DataFrame(data=fingerprints["moseq"], index=index)
moseq_df.index = pd.MultiIndex.from_tuples(moseq_df.index, names=index_levels)

moseq_df
```

```python
with open(data_root_dir / "dataset.pkl", "rb") as f:
    dataset = pickle.load(f, encoding='latin1')

def _dict_to_df(data_dict, index_key="mouse_img_idx", ignore_keys=("good_idx", "roi", "data", "pca_components")):
    return pd.DataFrame({col: data_dict[col].flatten() for col in data_dict.keys() if col not in (index_key, *ignore_keys)})

full_df = pd.concat({k: _dict_to_df(v) for k,v in dataset.items()}, names=["mouse_names", "frame_idx"])
full_df["velocity"] = full_df["velocity"].abs()  # why do negative velocities exist...?
full_df
```

```python
bin_dict = {
    "dist_to_center": np.linspace(0, 1, 90),
    "height": np.linspace(0, 60, 45),
    "velocity": np.linspace(0, 20, 90),
    "length": np.linspace(20, 100, 45)
}

def _dataset_to_scalars(dataset_df, density=False):
    if not density:
        hist_func = lambda x: pd.Series(np.concatenate([np.histogram(x[c], bins=bin_dict[c], density=False)[0] for c in bin_dict.keys()]))
    else:
        hist_func = lambda x: pd.Series(np.concatenate([np.histogram(x[c], bins=bin_dict[c], density=True)[0]/(1/np.diff(bin_dict[c])[1]) for c in bin_dict.keys()]))
    return dataset_df.groupby("mouse_names").apply(hist_func)

reproduced_scalars_df = _dataset_to_scalars(full_df, density=False)
_idx = pd.DataFrame(index, columns=index_levels).set_index("mouse_names")
reproduced_scalars_df = reproduced_scalars_df.join(_idx).reset_index().set_index(index_levels).reindex(index=scalars_df.index)
reproduced_scalars_df.equals(scalars_df)
```

```python
density_scalars_df = _dataset_to_scalars(full_df, density=True)
_idx = pd.DataFrame(index, columns=index_levels).set_index("mouse_names")
density_scalars_df = density_scalars_df.join(_idx).reset_index().set_index(index_levels).reindex(index=scalars_df.index)
print(density_scalars_df.equals(scalars_df))
density_scalars_df
```

```python
def preprocess_features(input_feature_df, n_components=0.95, rescale=True, pca=True):
    feature_array = input_feature_df.copy().values
    if rescale:
        feature_array = StandardScaler().fit_transform(feature_array)
    if pca:
        feature_array = PCA(n_components=n_components).fit_transform(feature_array)
    return pd.DataFrame(data=feature_array, index=input_feature_df.index)

pca_scalars_df = preprocess_features(density_scalars_df, rescale=False, pca=True)
pca_moseq_df = preprocess_features(moseq_df, rescale=False, pca=True)

rescaled_pca_scalars_df = preprocess_features(density_scalars_df, rescale=True, pca=True)
rescaled_pca_moseq_df = preprocess_features(moseq_df, rescale=True, pca=True)
```

```python
scalars_df_dict = dict(
    raw=scalars_df,
    density=density_scalars_df,
    pca=pca_scalars_df,
    rescaled_pca=rescaled_pca_scalars_df
)

approach_scalars_df = pd.concat(scalars_df_dict, names=["approach"])

moseq_df_dict = dict(
    raw=moseq_df,
    density=moseq_df,
    pca=pca_moseq_df,
    rescaled_pca=rescaled_pca_moseq_df
)

approach_moseq_df = pd.concat(moseq_df_dict, names=["approach"])

save_object(approach_scalars_df, data_output_dir / "approach_scalars_df.pkl", overwrite=True)
save_object(approach_moseq_df, data_output_dir / "approach_moseq_df.pkl", overwrite=True)
approach_scalars_df
```

```python
vector_df = pd.concat({
    "scalars": approach_scalars_df,
    "moseq": approach_moseq_df
}, names=["segmenter"]).sort_index()
vector_df.index.names = ["segmenter", "approach", "drug_class", "treatment", "uniform_dosage_term", "mouse_names"]
vector_df.columns.name = "behavior_name"
_idx = vector_df.index.to_frame(index=False)
_idx["uniform_dosage_term"] = _idx["uniform_dosage_term"].str.lower()
_idx["treatment_mouse_id"] =  _idx["mouse_names"].str.extract(r"^.+?\-(\d+)\_").astype(int)
_idx["phase"] = "Solitary"

_control_idx = _idx[_idx["treatment"].eq("control")]
_idx.loc[_control_idx.index, "treatment_mouse_id"] = _control_idx.groupby(["approach", "segmenter", "treatment", "uniform_dosage_term"]).apply(lambda x: pd.Series(range(len(x)), dtype=int), include_groups=False).stack().values
_idx.loc[_control_idx.index, "treatment"] = "VEH"
_idx.loc[_control_idx.index, "uniform_dosage_term"] = "medium"

vector_df.index = pd.MultiIndex.from_frame(_idx[["approach", "segmenter", "phase", "treatment", "uniform_dosage_term", "treatment_mouse_id"]])

save_object(vector_df.fillna(0), data_output_dir / "approach_vector_df.pkl", overwrite=True)
vector_df
```
